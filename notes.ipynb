# 📄 Custom Exception Class

```python
class InvalidURLException(Exception):
    """Exception raised for invalid URLs."""
    
    def __init__(self, message="The provided URL is invalid."):
        self.message = message
        super().__init__(self.message)
```

---

# 🧠 Key Concept

The class inherits from:

```
Exception
```

This makes it a **valid Python error type**.

Example usage:

```python
raise InvalidURLException("URL must start with https://")
```

---

# ⚙️ Custom Message Support

Constructor:

```python
def __init__(self, message="The provided URL is invalid."):
```

Allows custom messages:

```
InvalidURLException("URL must start with https")
```

---

# 🧪 Example Usage

```python
from nexusview.custom_exception import InvalidURLException

def validate_url(url):
    if not url.startswith("https"):
        raise InvalidURLException("URL must start with https")
```

Handling the exception:

```python
try:
    validate_url(url)
except InvalidURLException as e:
    logger.error(e)
```

# 🔎 Extracting the Video ID (Regex)

```python
regex = r"(?:v=|\/)([0-9A-Za-z_-]{11}).*"
match = re.search(regex, url)
```

Purpose:

```
Extract the YouTube video ID from the URL.
```

Example URL:

```
https://www.youtube.com/watch?v=dQw4w9WgXcQ
```

Extracted video ID:

```
dQw4w9WgXcQ
```

---

# 🎯 Capture Groups

Parentheses in regex create **capture groups**.

```
([0-9A-Za-z_-]{11})
```

This means:

```
capture 11-character YouTube video ID
```

Then:

```python
video_id = match.group(1)
```

returns:

```
dQw4w9WgXcQ
```

---

# ⚠️ Invalid URL Handling

If regex fails:

```python
if not match:
    raise InvalidURLException(f"Invalid YouTube URL: {url}")
```

Custom exception is raised.

Example error:

```
InvalidURLException: Invalid YouTube URL
```

---

# 🔗 Creating Embed URL

```python
embed_url = f"https://www.youtube-nocookie.com/embed/{video_id}"
```

This URL is used to **embed the YouTube player**.
